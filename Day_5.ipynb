```python
# sliding window using 2 pointers
l1=list(map(int,input().split()))
k=10
r,l,s,m=0,0,0,0
while r<len(l1):
    s+=l1[r]
    while s>=k:
        s-=l1[l]
        l+=1
    length=r-l+1
    m=max(m,length)
    r+=1
print(m)
```

```python
#leetcode 713 (subarray product less than k)
class Solution:
    def numSubarrayProductLessThanK(self, nums: List[int], k: int) -> int:
        if k<=1:
            return 0
        r,l=0,0
        cnt,p=0,1
        while (r<len(nums)):
            p*=nums[r]
            while p>=k:
                p//=nums[l]
                l+=1
            cnt+=r-l+1
            r+=1
        return cnt
```

```python
#leetcode 1423 (maximun points you can obtain from cards)
class Solution:
    def maxScore(self, cardPoints: List[int], k: int) -> int:
     n=len(cardPoints)
     s=sum(cardPoints[:k])
     m=s
     for i in range(k):
        s=s-cardPoints[k-1-i]+cardPoints[n-1-i]
        m=max(m,s)
     return m
```

```python
#leetcode 703 binary search
class Solution:
    def search(self, nums: List[int], target: int) -> int:
        low=0
        high=len(nums)-1
        while low<=high:
            mid=(low+high)//2
            if target<nums[mid]:
               high=mid-1
            elif target>nums[mid]:
                low=mid+1
            else :
                return mid
        return -1

```

```python
#leetcode 35 (search insert positon)
class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:

        low=0
        high=len(nums)-1
        while low<=high:
            mid=(low+high)//2
            if target<nums[mid]:
               high=mid-1
            elif target>nums[mid]:
                low=mid+1
            else :
                return mid
        return low
```

```python
#leetcode 34
class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        left=0
        right=len(nums)-1
        l=[]
        r=-1
        while(left<=right):
            mid=(left+right)//2
            if(nums[mid]==target):
                r=mid
                right=mid-1
            elif target<nums[mid]:
                right=mid-1
            else:
                left=mid+1

        l.append(r)
        r=-1
        left=0
        right=len(nums)-1
        while(left<=right):
            mid=(left+right)//2
            if(nums[mid]==target):
                r=mid
                left=mid+1
            elif target<nums[mid]:
                right=mid-1
            else:
                left=mid+1
        l.append(r)
        return l
```

```python
#leetcode 162
class Solution:
    def findPeakElement(self, nums: List[int]) -> int:
        n=len(nums)-1
        if n==0:
            return 0
        elif nums[0]>nums[1]:
            return 0
        elif nums[-1]>nums[-2]:
            return n
        s,e=0,n-1
        while s<=e:
            m=(s+e)//2
            if(nums[m-1])<nums[m]>nums[m+1]:
                return m
            elif nums[m]<nums[m+1]:
                s=m+1
            else:
                e=m-1
```

```python
#leetcode 74
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        row=len(matrix)
        col=len(matrix[0])
        s,e=0,(row*col)-1
        while s<=e:
            m=(s+e)//2
            i=m//col
            j=m%col
            if matrix[i][j]==target:
                return True
            elif matrix[i][j]<target:
                s=m+1
            else:
                e=m-1
        return False
```

```python
#leetcode 69 (sqrt)
class Solution:
    def mySqrt(self, x: int) -> int:
        low=0
        high=x
        while low<=high:
            mid=(low+high)//2
            if(mid*mid)==x:
                return mid
            elif (mid*mid)<x and((mid+1)*(mid+1))>x:
                return mid
            elif (mid*mid)<x:
                low=mid+1
            else:
                high=mid-1
```
