```python
#leetcode 79 (word search)//backtracking
class Solution:
    def exist(self, board: List[List[str]], word: str) -> bool:
        def possible(board,i,j,k):
            if k==len(word):
                return True
            if i<0 or i>=len(board) or j<0 or j>=len(board[0]) or board[i][j]!=word[k]:
                return False
            temp=board[i][j]
            board[i][j]='#'
            if possible(board,i-1,j,k+1) or possible(board,i+1,j,k+1) or possible(board,i,j-1,k+1) or possible(board,i,j+1,k+1):
                return True
            board[i][j]=temp
            return False
        for i in range(len(board)):
            for j in range(len(board[0])):
                if possible(board,i,j,0):
                    return True
        return False
```

```python
#leetcode 20 valid parentheses
class Solution:
    def isValid(self, s: str) -> bool:
        st=[]
        for i in range(len(s)):
            if s[i]=='(':
                st.append(')')
            elif s[i]=='{':
                st.append('}')
            elif s[i]=='[':
                st.append(']')
            elif len(st)==0 or st.pop()!=s[i]:
                return False
        return len(st)==0
```


#leetcode 844

class Solution:
    def backspaceCompare(self, s: str, t: str) -> bool:
        st=[]
        for i in s:
            if i!='#':
                st.append(i)
            elif len(st)>0:
                st.pop()
        sm=[]
        for c in t:
            if c!='#':
                sm.append(c)
            elif len(sm)>0:
                sm.pop()
        if st==sm:
            return True
        return False

```python
#leetcode 1700
class Solution:
    def countStudents(self, students: List[int], sandwiches: List[int]) -> int:
        n=len(students)
        while students and sandwiches and sandwiches[0] in students:
            if students[0]!=sandwiches[0]:
                students.append(students.pop((0)))
            else:
                n-=1
                students.pop(0)
                sandwiches.pop(0)
        return n
```

```python
#leetcode 496
class Solution:
    def nextGreaterElement(self, nums1: List[int], nums2: List[int]) -> List[int]:

            d={}
            st=[]
            for curr in nums2:
                while st and st[-1]<curr:
                    ele=st.pop()
                    d[ele]=curr
                st.append(curr)
            while st:
                d[st.pop()]=-1
            res=[]
            for i in nums1:
                res.append(d[i])
            return res
```
